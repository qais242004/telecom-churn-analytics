```python
import pandas as pd

df = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")

print(df.shape)
df.info()
print(df.isnull().sum())
print(df.duplicated().sum())
print(df["Churn"].value_counts(normalize=True))
```

```python
blank_rows = df[df["TotalCharges"].str.strip() == ""]
blank_rows[["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]]
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"].str.strip(), errors="coerce")
print("Blank values after converting:", df["TotalCharges"].isnull().sum())

df["TotalCharges"] = df["TotalCharges"].fillna(0)
print("Blank values after filling:", df["TotalCharges"].isnull().sum())
print("Data type now:", df["TotalCharges"].dtype)
```

```python
df["SeniorCitizen"] = df["SeniorCitizen"].map({0: "No", 1: "Yes"})

print(df["SeniorCitizen"].value_counts())
print("Missing after change:", df["SeniorCitizen"].isnull().sum())
```

```python
df.to_csv("../data/cleaned/telco_churn_cleaned.csv", index=False)

check = pd.read_csv("../data/cleaned/telco_churn_cleaned.csv")
print("Saved shape:", check.shape)
print(check["TotalCharges"].dtype)
print(check["SeniorCitizen"].unique())
```
